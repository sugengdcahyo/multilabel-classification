```python
import re
import string
import emoji
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def clean_text(text):
    # Hilangkan emoji dengan emoji library
    text = emoji.demojize(text)

    # Hilangkan tanda baca
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Hilangkan angka
    text = re.sub(r'\d+', '', text)

    # Hilangkan karakter khusus
    text = re.sub(r'\W+', ' ', text)

    # Ubah ke lowercase
    text = text.lower().strip()

    return text
```

```python

sample_text = "Hotel ini bagus banget!!! 😭 100% recommended!"
cleaned_text = clean_text(sample_text)
print(cleaned_text)
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk

nltk.download("stopwords")
nltk.download("punkt")

stop_words = set(stopwords.words("indonesian"))  # Bisa ditambah dengan stopwords khusus

def remove_stopwords(text):
    words = word_tokenize(text)
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)

# Contoh
cleaned_text_no_stopwords = remove_stopwords(cleaned_text)
print(cleaned_text_no_stopwords)  # Output: "hotel bagus banget love recommended"
```
